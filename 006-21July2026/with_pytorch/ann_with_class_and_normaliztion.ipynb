```python
# ==========================================
# 1. IMPORT LIBRARIES
# ==========================================

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
```

```python
# ==========================================
# 2. LOAD MNIST DATASET (WITH NORMALIZATION)
# ==========================================

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))   # <-- ADDED NORMALIZATION
])
```

```python
train_dataset = datasets.MNIST(
    root="./data",
    train=True,
    download=True, # download=False,
    transform=transform
)

test_dataset = datasets.MNIST(
    root="./data",
    train=False,
    download=True, # download=False,
    transform=transform
)

print(train_dataset)
print(test_dataset)
print(f"Number of training samples: {len(train_dataset)}")
print(f"Number of test samples: {len(test_dataset)}")
print(f"Shape of a single training sample: {train_dataset[0][0].shape}")
print(f"Label of a single training sample: {train_dataset[0][1]}")
print(f"Shape of a single test sample: {test_dataset[0][0].shape}")
print(f"Label of a single test sample: {test_dataset[0][1]}")
print(f"Data type of a single training sample: {train_dataset[0][0].dtype}")
print(f"Data type of a single test sample: {test_dataset[0][0].dtype}")
print(f"Minimum pixel value of a single training sample: {train_dataset[0][0].min()}")
print(f"Maximum pixel value of a single training sample: {train_dataset[0][0].max()}")
print(f"Minimum pixel value of a single test sample: {test_dataset[0][0].min()}")
print(f"Maximum pixel value of a single test sample: {test_dataset[0][0].max()}")
print(f"Mean pixel value of a single training sample: {train_dataset[0][0].mean()}")
print(f"Mean pixel value of a single test sample: {test_dataset[0][0].mean()}")
print(f"Standard deviation of pixel values of a single training sample: {train_dataset[0][0].std()}")
print(f"Standard deviation of pixel values of a single test sample: {test_dataset[0][0].std()}")
print(f"Shape of the training dataset: {train_dataset.data.shape}")
print(f"Shape of the test dataset: {test_dataset.data.shape}")
print(f"Data type of the training dataset: {train_dataset.data.dtype}")
print(f"Data type of the test dataset: {test_dataset.data.dtype}")
print(f"Minimum pixel value of the training dataset: {train_dataset.data.min()}")
print(f"Maximum pixel value of the training dataset: {train_dataset.data.max()}")
print(f"Minimum pixel value of the test dataset: {test_dataset.data.min()}")
print(f"Maximum pixel value of the test dataset: {test_dataset.data.max()}")
print(f"Mean pixel value of the training dataset: {train_dataset.data.float().mean()}")
print(f"Mean pixel value of the test dataset: {test_dataset.data.float().mean()}")
print(f"Standard deviation of pixel values of the training dataset: {train_dataset.data.float().std()}")
print(f"Standard deviation of pixel values of the test dataset: {test_dataset.data.float().std()}")
print(f"Number of classes in the training dataset: {len(set(train_dataset.targets.tolist()))}")
print(f"Number of classes in the test dataset: {len(set(test_dataset.targets.tolist()))}")
print(f"Classes in the training dataset: {set(train_dataset.targets.tolist())}")
print(f"Classes in the test dataset: {set(test_dataset.targets.tolist())}")
print(f"Number of samples per class in the training dataset: {torch.bincount(train_dataset.targets)}")
print(f"Number of samples per class in the test dataset: {torch.bincount(test_dataset.targets)}")
```

```python
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)
```

```python
# ==========================================
# 3. DEFINE MODEL (ANN)
# ==========================================

class MNIST_ANN(nn.Module):

    def __init__(self):
        super().__init__()

        self.model = nn.Sequential(

            nn.Linear(28*28, 128),
            nn.ReLU(),

            nn.Linear(128, 64),
            nn.ReLU(),

            nn.Linear(64, 10)
        )

    def forward(self, x):
        print(f"Input shape - inside forward: {x.shape}")
        x = x.view(-1, 28*28)
        return self.model(x)


model = MNIST_ANN()
```

```python

# ==========================================
# 4. LOSS & OPTIMIZER
# ==========================================

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python
# ==========================================
# 5. TRAINING LOOP
# ==========================================

epochs = 5

for epoch in range(epochs):

    model.train()
    total_loss = 0

    for images, labels in train_loader:

        print(f"forward method in the class should be called now - model(images) is calling the forward method")
        outputs = model(images)
        print(f"forward has been called, output shape: {outputs.shape}")

        loss = criterion(outputs, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch [{epoch+1}/{epochs}], Loss: {total_loss:.4f}")
```

```python
# ==========================================
# 6. TESTING
# ==========================================

model.eval()

correct = 0
total = 0

with torch.no_grad():

    for images, labels in test_loader:

        outputs = model(images)

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = correct / total

print("\nTest Accuracy:", accuracy)
```
